```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('TSLA.csv')
data.head()
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.info()
```

```python
data = data.sort_values(by='Date')
```

Average closing price
The closing price is the final price at which Tesla stock traded at the end of each trading day. 

```python
average_close = data['Close'].mean()
average_close
```

```python
lowest_close = data['Close'].min()
lowest_close
```

```python
highest_close = data['Close'].max()
highest_close
```

```python
print(f"The average closing price of Tesla stock is: ${average_close:.2f}")
print(f"The highest closing price of Tesla stock is: ${highest_close:.2f}")
print(f"The lowest closing price of Tesla stock is: ${lowest_close:.2f}")
```

Highest and lowest Closing_Price Dates
Highest closing-price row

```python
highest_close_index = data['Close'].idxmax()
data.loc[highest_close_index]
```

```python
lowest_close_index = data['Close'].idxmin()
data.loc[lowest_close_index]
```

VISUALIZE TESLA'S CLOSING PRICE
Tesla closing price over time
The following line chart shows how Tesla's closing stock price changed during the period covered by dataset.

```python
plt.figure(figsize=(12, 6))
plt.plot(data['Date'], data['Close'], label='Closing Price', color='blue')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.title('Tesla Stock Price')
plt.legend()
plt.show()
```

The line graph shows that Tesla's closing price experienced several periods of growth and decline. This indicate that the stock was volatile rather tahn remaining stable throughout the period.

```python
data['Daily Change'] = data['Close'] - data ['Open']
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data['Date'], 
         data['Daily Change'], 
         color='purple', 
         label='Daily price Change')

plt.axhline(y=0, color='black', linestyle='--', linewidth=1)
plt.title('Tesla Daily Price Change', fontsize=16)
plt.xlabel('Date', fontsize=14)
plt.ylabel('Daily Price Change', fontsize=14)
plt.grid(True, alpha =0.3)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Interpretation
Values above zero represent days when Tesla closed higher than it opened. values below zero represent days when Tesla closed lower than it opened.
The graph shows that Tesla experienced both positive and negative daily price movements, confirming that the stock fluctuated throughout the period.

```python
average_volume = data['Volume'].mean()
highest_volume = data['Volume'].max()
print(f" average volume of Tesla stock is, {average_volume}")
print(f" highest volume of Tesla stock is, {highest_volume}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data['Date'], data['Volume'], color='orange', label='Trading Volume')
plt.title('Tesla Trading Volume Over Time', fontsize=16)
plt.xlabel('Date', fontsize=14)
plt.ylabel('Trading Volume', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
